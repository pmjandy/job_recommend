```python
import requests
import pandas as pd
from bs4 import BeautifulSoup
```

```python
from google.colab import drive
drive.mount('/content/drive')
df = pd.read_csv('/content/drive/MyDrive/bigdata/커리어넷_직업백과.csv')
```

```python
key = '29c85897053f2d7e2ea35c3820b18086'
start = 1
last = 61

url = f"https://www.career.go.kr/cnet/front/openapi/jobs.json?apiKey={key}"
params = {"pageSize": 10, "pageIndex": 1}
r = requests.get(url, params = params)

data = r.json()
data
```

```python
count = data.get("count")

if count:
  total_pages = (count + params["pageSize"]-1) // params['pageSize']
```

```python
jobs = data.get("jobs", [])

all_jobs = jobs[:]
current_page = params["pageIndex"]

for p in range(current_page+1, total_pages+1):
  params["pageIndex"] = p
  page_data = r.json()
  page_jobs = page_data.get("jobs",[])
  if not page_jobs:
    break
  all_jobs.extend(page_jobs)

print("총 직업: ", len(all_jobs))
```

```python
information_ls = []

for i in all_jobs:
  information_dict = {}
  information_dict["직업명"] = i["job_nm"]
  information_dict["직업코드"] = i["seq"]
  information_dict["업무"] = i["work"]
  information_ls.append(information_dict)

information_ls
```

```python
table = pd.DataFrame.from_dict(information_ls, orient = "columns")
table
```

```python
path = "/content/커리어넷_직업백과.csv"
table.to_csv(path, index = False, encoding = "utf-8-sig")
```

```python
jobcode = df['직업코드'].tolist()
jobcode
```

```python
all_details = []

for code in jobcode:
  url_details = f'https://www.career.go.kr/cnet/front/openapi/job.json?apiKey={key}&seq={code}'
  r2 = requests.get(url_details)
  detail = r2.json()
  all_details.append(detail)

all_details
```

```python
detail_ls = []

for d in all_details:
  detail_dict = {}

  detail_dict["직업코드"] = d["baseInfo"]["job_cd"]

  detail_dict["직업명"] = d["baseInfo"]["job_nm"]

  detail_dict["직업분류"] = d["baseInfo"]["aptit_name"]

  detail_dict["하는일"] = d["workList"]

  detail_dict["핵심능력"] = d["abilityList"]

  detail_dict["관련학과"] = d["departList"]

  detail_dict["적성"] = d["aptitudeList"]

  detail_dict["흥미"] = d["interestList"]

  detail_dict["직업전망"] = d["forecastList"]

  detail_dict["업무환경"] = d["performList"]["environment"]

  detail_dict["업무수행"] = d["performList"]["perform"]

  detail_ls.append(detail_dict)

detail_ls
```

```python
table = pd.DataFrame.from_dict(detail_ls, orient = "columns")
table
```

```python
path = "/content/커리어넷_최최종.csv"
table.to_csv(path, index = False, encoding = "utf-8-sig")
```
